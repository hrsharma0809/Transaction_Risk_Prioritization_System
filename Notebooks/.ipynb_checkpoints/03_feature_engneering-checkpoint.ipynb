```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('Synthetic_Transaction_Risk_Dataset.xlsx')
```

```python
df.columns
```

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'])
```

```python
df['DayOfWeek'] = df['Timestamp'].dt.day_name()
```

```python
df['Month'] = df['Timestamp'].dt.month
```

```python
df['IsNightTransaction'] = (
    (df['TransactionHour'] >= 0) &
    (df['TransactionHour'] <= 5)
).astype(int)
```

```python
threshold = df['Amount'].quantile(0.95)

df['HighValueTxn'] = (df['Amount'] > threshold).astype(int)
```

```python
df['AmountRatio'] = (
    df['Amount'] /
    df['AvgCustomerAmount']
)
```

```python
df['HighRiskMerchant'] = (
    df['MerchantRiskScore'] > 0.8
).astype(int)
```

```python
df['NewCustomer'] = (
    df['AccountAgeDays'] < 30
).astype(int)
```

```python
df['OldCustomer'] = (
    df['AccountAgeDays'] > 365
).astype(int)
```

```python
df['AbsoluteDeviation'] = abs(df['AmountDeviation'])
```

```python
from sklearn.preprocessing import MinMaxScaler

cols = [
    'AmountRatio',
    'MerchantRiskScore',
    'PrevTxn24H',
    'AbsoluteDeviation'
]

scaler = MinMaxScaler()

scaled = scaler.fit_transform(df[cols])

df['CompositeRiskScore'] = scaled.mean(axis=1)
```

```python
df.columns
```

```python
df= df.drop(columns = ['TransactionID', 'Timestamp', 'CustomerID', 'MerchantID'])
```

```python
df.columns
```

```python
df['City'].nunique()
```

```python
categorical_cols = [
    'MerchantCategory',
    'City',
    'State',
    'PaymentMethod',
    'DeviceType',
    'DayOfWeek'
]

df = pd.get_dummies(
    df,
    columns=categorical_cols,
    drop_first=True
)
```

```python
corr = df.corr(numeric_only=True)
```
